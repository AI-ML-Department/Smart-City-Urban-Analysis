```python
numeric_cols = df.select_dtypes(include='number').columns
print("Numeric columns:", numeric_cols)
categorical_cols = df.select_dtypes(include='str').columns
print("Categorical columns:", categorical_cols)
```

```python
df.duplicated().sum()
```

```python
for col in categorical_cols:
    print(f"Value counts for {col}:")
    print(df[col].value_counts())
```

```python
for col in numeric_cols:
    q1=df[col].quantile(0.25)
    q3=df[col].quantile(0.75)
    iqr=q3-q1
    lower_bound=q1-1.5*iqr
    upper_bound=q3+1.5*iqr
    outliers=df[(df[col]<lower_bound) | (df[col]>upper_bound)]
    print(f"Outliers for {col}: {outliers.shape[0]}")
```

```python
#part 4
#age distribution histogram
sns.histplot(df['Age'])
plt.title('Age Distribution')
plt.show()
```

```python
#part 5
irr_col=['Citizen_ID','Calories_Burned','Sleep_Hours','Home_Energy_Consumption_kWh','Social_Media_Hours']
df_relevant=df.drop(columns=irr_col)
df_relevant.head(5)
```

```python
df_encoded = pd.get_dummies(df_relevant,
                            columns=["Mode_of_Transport","Gender"],
                            drop_first=True,
                            dtype=int)
df_encoded.describe()
```

```python
col=['Age','Work_Hours','Shopping_Hours','Entertainment_Hours','Carbon_Footprint_kgCO2','Steps_Walked','Public_Events_Hours']
df_standardized = df_encoded.copy()
for c in col:
    df_standardized[c] = (df_encoded[c] - df_encoded[c].mean()) / df_encoded[c].std()

df_standardized.describe().round(2)
```

```python
df_standardized.to_csv('smart_city_cleaned.csv.csv', index=False)
```
